```python
import requests
import pandas as pd
```

# **Scrapping player data (Messi)**

```python
url = "http://footballapi.eastus.azurecontainer.io:3000/players/search?q=messi&page=15"
```

```python
response = requests.get(url)
data = response.json()
```

```python
print(data)
```

```python
df = pd.DataFrame(data)
df
```

```python
df_flat = pd.json_normalize(df['entity'])
df_flat['score'] = df['score']
df_flat['type'] = df['type']
df_flat
```

```python
match_info = {
    "home_formation": data4['home'].get('formation'),
    "away_formation": data4['away'].get('formation'),
    "confirmed": data4['confirmed']
}
print(match_info)
```
